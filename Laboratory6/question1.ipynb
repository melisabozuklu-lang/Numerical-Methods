# Exercise 1 – Aitken Acceleration for Newton’s Method

In this exercise, Newton’s method was used to solve a nonlinear system of equations. Newton’s method is one of the most common numerical methods for solving nonlinear equations because of its fast convergence near the exact solution. However, in some cases the convergence can become slow, especially when the starting approximation is not sufficiently close to the root or when the system behaves poorly.

To improve the convergence speed, Aitken acceleration can be applied. The main idea of Aitken’s method is to use consecutive approximations obtained during the iteration process and estimate a new approximation that converges faster to the exact solution.

The Aitken acceleration formula is given by:

$$
x_k^* = x_k - \frac{(x_{k+1}-x_k)^2}{x_{k+2}-2x_{k+1}+x_k}
$$

where:

- xk represents the current approximation,
- x(k+1) and x(k+2) represent the following approximations,
- x*k represents the accelerated approximation obtained using Aitken’s method.

For multidimensional systems, the acceleration process is applied component-wise to each variable separately.

In this exercise, the provided code in the file `1.ipynb` was used. First, the nonlinear system was solved using the standard Newton’s method. After that, Aitken acceleration was applied to the iterative sequence in order to improve convergence.

The obtained results showed that the accelerated method converged faster than the standard Newton’s method. The number of iterations required to reach the solution decreased, and the approximations stabilized more quickly.

Therefore, Aitken acceleration can be considered an effective technique for improving slowly converging Newton iterations and increasing computational efficiency in numerical methods.
