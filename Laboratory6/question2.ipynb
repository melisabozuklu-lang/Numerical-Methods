# Exercise 2 – Gaussian Substitution

In this exercise, the given system of linear equations was solved manually using Gaussian substitution.

The system is:

$$
4x_1 - x_2 + x_3 = 8
$$

$$
2x_1 + 5x_2 + 2x_3 = 3
$$

$$
x_1 + 2x_2 + 4x_3 = 11
$$

First, the equations were written in augmented matrix form:

$$
\begin{bmatrix}
4 & -1 & 1 & | & 8 \\
2 & 5 & 2 & | & 3 \\
1 & 2 & 4 & | & 11
\end{bmatrix}
$$

To eliminate \(x_1\) from the second and third equations, the following row operations were applied:

$$
R_2 \rightarrow R_2 - 0.5R_1
$$

$$
R_3 \rightarrow R_3 - 0.25R_1
$$

After these operations, the system becomes:

$$
\begin{bmatrix}
4 & -1 & 1 & | & 8 \\
0 & 5.5 & 1.5 & | & -1 \\
0 & 2.25 & 3.75 & | & 9
\end{bmatrix}
$$

Next, \(x_2\) was eliminated from the third equation:

$$
R_3 \rightarrow R_3 - \frac{2.25}{5.5}R_2
$$

This gives the triangular system:

$$
\begin{bmatrix}
4 & -1 & 1 & | & 8 \\
0 & 5.5 & 1.5 & | & -1 \\
0 & 0 & 3.136 & | & 9.409
\end{bmatrix}
$$

Using back substitution, the solution is obtained as:

$$
x_3 = 3
$$

$$
x_2 = -1
$$

$$
x_1 = 1
$$

Therefore, the final solution is:

$$
x_1 = 1,\quad x_2 = -1,\quad x_3 = 3
$$
