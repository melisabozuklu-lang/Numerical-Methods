# Exercise 1 – Numerical Integration of ∫₁⁴ x² dx

## Exact Value

The analytical value of the integral is:

∫₁⁴ x² dx = [x³/3]₁⁴ = (64 − 1)/3 = 21

---

## (a) Left Rectangle Method

Δx = 1

Using the left endpoints:

Aₗ = 1 × (1 + 4 + 9)

Aₗ = 14

Error:

|14 − 21| = 7

---

## (b) Right Rectangle Method

Δx = 1

Using the right endpoints:

Aᵣ = 1 × (4 + 9 + 16)

Aᵣ = 29

Error:

|29 − 21| = 8

---

## (c) Midpoint Rectangle Method

Midpoints:

1.5, 2.5, 3.5

Function values:

f(1.5) = 2.25

f(2.5) = 6.25

f(3.5) = 12.25

Aₘ = 1 × (2.25 + 6.25 + 12.25)

Aₘ = 20.75

Error:

|20.75 − 21| = 0.25

---

## (d) Trapezoidal Method

Aₜ = (Δx / 2)[f(1) + f(4) + 2(f(2) + f(3))]

Aₜ = (1 / 2)[1 + 16 + 2(4 + 9)]

Aₜ = (1 / 2)(43)

Aₜ = 21.5

Error:

|21.5 − 21| = 0.5

---

## Results

| Method             | Approximation | Error |
| ------------------ | ------------- | ----- |
| Left Rectangle     | 14.00         | 7.00  |
| Right Rectangle    | 29.00         | 8.00  |
| Midpoint Rectangle | 20.75         | 0.25  |
| Trapezoidal        | 21.50         | 0.50  |

## Conclusion

The midpoint rectangle method provides the most accurate approximation with an error of 0.25. The trapezoidal method is also very accurate with an error of 0.5. The left and right rectangle methods produce significantly larger errors.
