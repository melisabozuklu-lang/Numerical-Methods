# Exercise 3 – Numerical Integration of ∫₋₁³ exp(-x²) dx

The function

f(x) = exp(-x²)

was numerically integrated on the interval [-1, 3] using four numerical methods.

The exact value calculated using SciPy is:

Actual area = 1.633031

## Results

| Method             | Approximation | Error (%) |
| ------------------ | ------------- | --------- |
| Left Rectangle     | 1.702554      | 4.2573    |
| Right Rectangle    | 1.539107      | 5.7516    |
| Midpoint Rectangle | 1.639161      | 0.3754    |
| Trapezoidal        | 1.620830      | 0.7471    |

## Discussion

The midpoint rectangle method produced the most accurate result with the smallest error (0.3754%). The trapezoidal method also provided a very accurate approximation with an error below 1%.

The left and right rectangle methods were significantly less accurate, producing errors above 4%.

The precision of numerical integration depends on both the selected method and the number of nodes. Increasing the number of nodes improves the approximation because the numerical model follows the function more closely. For this function, the midpoint rectangle and trapezoidal methods provide the best accuracy.


```
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate

def f(x):
    return np.exp(-x**2)

actual = integrate.quad(lambda x: f(x), -1, 3)[0]

def draw_method(a,b,points,method):

    x = np.linspace(a,b,1000)
    plt.figure(figsize=(8,4))
    plt.plot(x,f(x),'r')

    n = np.linspace(a,b,points)

    if method=="rectangularL":
        for i in range(len(n)-1):
            plt.bar(n[i],f(n[i]),width=n[i+1]-n[i],
                    align='edge',fill=False,edgecolor='blue')

    elif method=="rectangularR":
        for i in range(len(n)-1):
            plt.bar(n[i],f(n[i+1]),width=n[i+1]-n[i],
                    align='edge',fill=False,edgecolor='blue')

    elif method=="rectangularM":
        for i in range(len(n)-1):
            m=(n[i]+n[i+1])/2
            plt.bar(n[i],f(m),width=n[i+1]-n[i],
                    align='edge',fill=False,edgecolor='blue')

    elif method=="trapezoidal":
        for i in range(len(n)-1):
            plt.plot([n[i],n[i]],[0,f(n[i])],'b')
            plt.plot([n[i+1],n[i+1]],[0,f(n[i+1])],'b')
            plt.plot([n[i],n[i+1]],[f(n[i]),f(n[i+1])],'b')

    plt.title(method)
    plt.show()

draw_method(-1,3,10,"rectangularL")
draw_method(-1,3,10,"rectangularR")
draw_method(-1,3,10,"rectangularM")
draw_method(-1,3,10,"trapezoidal")
```
